# Clips YouTube Downloader (v1)
Paste the YouTube link in **URL**, then Runtime > Run all. The video lands in My Drive/AI/Clips Channel/Sources. Clips channel only.

```
URL = "" #@param {type:"string"}
MAX_HEIGHT = "1080" #@param ["720", "1080", "1440", "2160"]
from google.colab import drive
drive.mount('/content/drive')
OUT = '/content/drive/MyDrive/AI/Clips Channel/Sources'
!mkdir -p "{OUT}"
!pip install -q -U yt-dlp
FMT = f'bv*[height<={MAX_HEIGHT}][ext=mp4]+ba[ext=m4a]/b[height<={MAX_HEIGHT}][ext=mp4]/b'
!yt-dlp --no-overwrites --no-playlist -f "{FMT}" --merge-output-format mp4 -o "{OUT}/%(title).80s [%(id)s].%(ext)s" "{URL}"
print('DONE_CLIPS_DOWNLOAD')
```
